```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.preprocessing import image
import matplotlib.pyplot as plt
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout


base_model = ResNet50(weights=None, include_top=False, input_shape=(224, 224, 3))

model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(256, activation='relu'),
    Dropout(0.3),
    Dense(4, activation='softmax') 
])


model.load_weights('alzheimers_resnet50_v1.keras')


class_names = ['MildDemented', 'ModerateDemented', 'NonDemented', 'VeryMildDemented']

def predict_brain(img_path):
    img = image.load_img(img_path, target_size=(224, 224))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) 
    img_array /= 255.0 

    predictions = model.predict(img_array)
    predicted_index = np.argmax(predictions[0])
    
    predicted_class = class_names[predicted_index]
    confidence = predictions[0][predicted_index] * 100

    plt.imshow(img)
    plt.title(f"AI Prediction: {predicted_class}\nConfidence: {confidence:.2f}%")
    plt.axis('off')
    plt.show()


predict_brain('img1.jpg')
```

```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_and_save_transfer_learning(csv_path1, csv_path2, save_name="results.png"):
    phase1 = pd.read_csv(csv_path1)
    phase2 = pd.read_csv(csv_path2)

    fig, axs = plt.subplots(2, 2, figsize=(16, 12))
    
    # Phase 1
    axs[0, 0].plot(phase1['accuracy'], color='blue', label='Train')
    axs[0, 0].plot(phase1['val_accuracy'], color='orange', label='Val')
    axs[0, 0].set_title('Phase 1: Accuracy')
    
    axs[0, 1].plot(phase1['loss'], color='blue')
    axs[0, 1].plot(phase1['val_loss'], color='orange')
    axs[0, 1].set_title('Phase 1: Loss')

    # Phase 2
    axs[1, 0].plot(phase2['accuracy'], color='green', label='Train')
    if 'val_accuracy' in phase2.columns:
        axs[1, 0].plot(phase2['val_accuracy'], color='red', label='Val')
    
    axs[1, 1].plot(phase2['loss'], color='green')
    if 'val_loss' in phase2.columns:
        axs[1, 1].plot(phase2['val_loss'], color='red')
        
    axs[1, 0].set_title('Phase 2: Accuracy')
    axs[1, 1].set_title('Phase 2: Loss')

    for ax in axs.flat:
        ax.set(xlabel='Epochs', ylabel='Value')
        ax.grid(True, alpha=0.3)
        ax.legend()

    plt.tight_layout()
    
    # This saves the file to your computer/current folder
    plt.savefig(save_name, dpi=300, bbox_inches='tight')
    print(f"Plot saved as {save_name}")
    
    plt.show()

# Run the function
path1 = r"C:\Users\NEW\ArnavDS\ALZ_DECT_TRANSFERLEARNING\clinical_results\content\my_dementia_model_results\training_log_phase1.csv"
path2 = r"C:\Users\NEW\ArnavDS\ALZ_DECT_TRANSFERLEARNING\clinical_results\content\my_dementia_model_results\training_log_phase2.csv"

plot_and_save_transfer_learning(path1, path2, "EEG_Transfer_Learning_Results.png")
```
